# TF-IDF feature extraction example

Course-provided example, adapted to the repository layout. Run from the project root or the `notebooks/` directory. This notebook extracts features only; it does not train a classifier.

The original HTML export is archived in `docs/examples/` and retains its original paths and outputs.

#### Import the necessary libraries

```python
from pathlib import Path

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "data" / "raw").is_dir():
    PROJECT_ROOT = PROJECT_ROOT.parent
DATA_DIR = PROJECT_ROOT / "data" / "raw"
```

#### Read the data to pandas DataFrame

```python
train_df = pd.read_json(DATA_DIR / "train.json")
print(train_df.shape)
train_df.head()
```

#### You can use the TfidfVectorizer from the scikit-learn library to convert the text to vectors

The vectorizer should only be fitted on the training data as the test data are unknown during training. 

```python
vectorizer = TfidfVectorizer()
vectorizer = vectorizer.fit(train_df['reviews'])
X_train = vectorizer.transform(train_df['reviews'])
X_train.shape
```

#### Test set should be transformed using the same vectorizer (fitted on the training data)

```python
test_df = pd.read_json(DATA_DIR / "test.json")
print(test_df.shape)
X_test = vectorizer.transform(test_df['reviews'])
print(X_test.shape)
```
